```python
# Load the png 
import matplotlib.pyplot as plt
import matplotlib.image as mpimg  # for image reading 
import os



disk_name = "AA_Tau"
img_dir = os.path.join(r"D:\CPD_MPIA\Images", disk_name)
labels = ["clean", "SNR", "sigma", "resid", "fullfov"]

all_pngs = [f for f in os.listdir(img_dir)]
selected_files = []
for label in labels:
    for fname in all_pngs:
        if label.lower() in fname.lower():  # case insensitive
            selected_files.append(fname)
            break  


panel_labels = ["(a)", "(b)", "(c)", "(d)", "(e)"]
fig, axes = plt.subplots(2, 3, figsize=(24, 20))
for i, (fname, label) in enumerate(zip(selected_files, panel_labels)):
    row, col = divmod(i, 3)
    ax = axes[row, col]
    img = mpimg.imread(os.path.join(img_dir, fname))
    ax.imshow(img)
    ax.set_title(f"{label}", fontsize=16)
    ax.axis('off')
    axes[1, 2].axis('off')   # hide the last empty subplot 




plt.tight_layout()
plt.subplots_adjust(wspace=0, hspace=0)  
plt.show()
```
