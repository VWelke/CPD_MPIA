```python
import os
import glob

# Variables
target = 'AA_Tau'
gap_ix = 0  
flux = 'F24uJy'  # Adjust this to your actual flux value

# Clean up files for all suffixes 0000 to 0099
for i in range(100):  # 0000 to 0099
    file_suffix = f'_{flux}_{i:04d}'  # Creates _F24uJy_0000, _F24uJy_0001, etc.
    base_name = f'{target}_gap{gap_ix}{file_suffix}'
    
    files_found = False
    
    # Move files (FIXED INDENTATION)
    resid_file = target+'_gap'+str(gap_ix)+file_suffix + '_frank_uv_resid.npz'
    if os.path.exists(resid_file):
        os.rename(resid_file, os.path.join('resid_vis', resid_file))
        print(f"✓ Moved {resid_file}")
        files_found = True
    
    # Move profile fit file (FIXED INDENTATION)
    profile_file = target+'_gap'+str(gap_ix)+file_suffix + '_frank_profile_fit.txt'
    if os.path.exists(profile_file):
        os.rename(profile_file, os.path.join('mprofiles', profile_file))
        print(f"✓ Moved {profile_file}")
        files_found = True

    # Remove remaining frank files for this suffix
    frank_files = glob.glob(base_name + '_frank*')
    for file in frank_files:
        os.remove(file)
        print(f"✓ Removed {file}")
        files_found = True
    
    # Optional: print summary for suffixes with no files
    if not files_found:
        print(f"• No files found for suffix {i:04d}")

print(f"\nCleanup completed for all suffixes 0000-0099")
```
